```python
import pandas as pd
import numpy as np
```

```python
import random

np.random.seed(42)

rows = 1000

study_hours = np.random.randint(1, 11, rows)

attendance = np.random.randint(60, 101, rows)

assignments = np.random.randint(1, 11, rows)

quiz_score = np.random.randint(40, 101, rows)
```

```python
exam_marks = (
    study_hours * 4
    + attendance * 0.25
    + assignments * 2
    + quiz_score * 0.35
    + np.random.randint(-5, 6, rows)
)

exam_marks = np.round(exam_marks).astype(int)
```

```python
df = pd.DataFrame({

    "Study_Hours":study_hours,

    "Attendance":attendance,

    "Assignments":assignments,

    "Quiz_Score":quiz_score,

    "Exam_Marks":exam_marks

})

df.head()
```

```python
df.to_csv(
    "/content/student_performance.csv",
    index=False
)

print("Dataset Saved Successfully!")
```
